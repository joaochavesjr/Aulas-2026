```python
import torch
import torch.nn as nn
import torchvision.transforms as transforms
import torchvision.models as models
from torchvision.datasets import CIFAR10
import numpy as np
import matplotlib.pyplot as plt
from scipy import linalg
from PIL import Image
import warnings
warnings.filterwarnings('ignore')
print("🎨 Crítico de Arte Digital inicializado!")
print("🔬 Laboratório equipado com:")
print("   🧠 Inception-V3 (nosso 'olho treinado')")
print("   📊 Calculadora FID")
print("   ⭐ Medidor Inception Score")
print("   📈 Analisador de qualidade")
```

```python
print("👁️ PREPARANDO O 'OLHO TREINADO' (Inception-V3)")
print("=" * 50)
print("🤔 Por que Inception-V3?")
print("   • Treinado em milhões de imagens reais")
print("   • Reconhece 1000 categorias diferentes")
print("   • É como um crítico que viu toda a história da arte")
print("   • Suas 'opiniões' são estatisticamente confiáveis")
print()
# Carregando Inception-V3 pré-treinado
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
inception_model = models.inception_v3(pretrained=True, transform_input=False)
inception_model.eval()
inception_model = inception_model.to(device)
# Removendo a última camada para obter features
inception_model.fc = nn.Identity()
print(f"✅ Inception-V3 carregado no: {device}")
print("🎯 Configurado para extrair features (não classificar)")
print("📊 Pronto para analisar qualidade de imagens")
```

```python
# Transform específico para Inception-V3
inception_transform = transforms.Compose([
    transforms.Resize((299, 299)),  # Inception precisa de 299x299
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                        std=[0.229, 0.224, 0.225])  # ImageNet normalization
])
print("🔧 Pré-processamento configurado:")
print("   📐 Redimensionamento: 299x299 (padrão Inception)")
print("   🎨 Normalização: ImageNet (como o modelo foi treinado)")
print("   ✅ Pronto para receber imagens para análise")
```

```python
def extrair_features_inception(images, model, batch_size=32):
    """Extrai features usando Inception-V3 (como um crítico analisa arte)"""
    print(f"🔍 Analisando {len(images)} obras de arte...")

    model.eval()
    features_list = []

    # Processando em lotes
    for i in range(0, len(images), batch_size):
        batch = images[i:i+batch_size]
        with torch.no_grad():
            # Extraindo features (a 'opinião' do Inception)
            features = model(batch)
            features_list.append(features.cpu().numpy())
# Concatenando todas as features
    all_features = np.concatenate(features_list, axis=0)
    print(f"✅ Análise concluída! Shape das features: {all_features.shape}")
    return all_features
# Demonstrando com imagens reais do CIFAR-10
print("📚 Carregando galeria de referência (CIFAR-10)...")
dataset = CIFAR10(root='./data', train=True, download=True, transform=inception_transform)
dataloader = torch.utils.data.DataLoader(dataset, batch_size=32, shuffle=False)
# Pegando uma amostra para demonstração
sample_images = []
sample_labels = []
for i, (images, labels) in enumerate(dataloader):
    sample_images.append(images.to(device))
    sample_labels.extend(labels.numpy())
    if i >= 3:  # Apenas algumas amostras para demonstração
        break
sample_batch = torch.cat(sample_images, dim=0)

print(f"🎨 Amostra da galeria: {sample_batch.shape[0]} imagens")
# Analisando as "opiniões" do Inception:
# Extraindo features da amostra
print("\n🧠 EXTRAINDO 'OPINIÕES' DO INCEPTION")
print("=" * 40)
sample_features = extrair_features_inception(sample_batch, inception_model)
# Analisando as estatísticas
print("📊 ESTATÍSTICAS DAS FEATURES:")
print(f"   📏 Dimensionalidade: {sample_features.shape[1]} features por imagem")
print(f"   📈 Média geral: {sample_features.mean():.4f}")
print(f"   📊 Desvio padrão: {sample_features.std():.4f}")
print(f"   📉 Valor mínimo: {sample_features.min():.4f}")
print(f"   📈 Valor máximo: {sample_features.max():.4f}")
# Visualizando distribuição das features
plt.figure(figsize=(12, 4))
plt.subplot(1, 3, 1)
plt.hist(sample_features.flatten(), bins=50, alpha=0.7, color='blue')
plt.title('📊 Distribuição das Features')
plt.xlabel('Valor da Feature')
plt.ylabel('Frequência')
plt.subplot(1, 3, 2)
plt.plot(sample_features.mean(axis=0)[:100])  # Primeiras 100 features
plt.title('📈 Média das Features (primeiras 100)')
plt.xlabel('Índice da Feature')
plt.ylabel('Valor Médio')
plt.subplot(1, 3, 3)
plt.plot(sample_features.std(axis=0)[:100])   # Primeiras 100 features
plt.title('📊 Desvio Padrão das Features')
plt.xlabel('Índice da Feature')
plt.ylabel('Desvio Padrão')
plt.tight_layout()
plt.show()
```
