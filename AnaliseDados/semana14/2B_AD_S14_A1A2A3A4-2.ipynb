## Projeto de Data Storytelling:

### "Concentração de Medalhas nas Olimpiadas"

```python
df_athletes = pd.read_csv("./athletes.csv")
df_medals_total = pd.read_csv("./medals_total.csv")
```

### preparação dos dados

```python
# Contando o número de atletas por país e gênero
gender_count = df_athletes.groupby(['country', 'gender']).size().unstack(fill_value=0)

# Calculando o total de atletas por país
gender_count['total'] = gender_count.sum(axis=1)

# Calculando o percentual de cada gênero
gender_count_percent = gender_count.div(gender_count['total'], axis=0) * 100

# Criando a tabela final com o total de atletas e os percentuais por gênero
gender_count_final = gender_count.drop(columns='total')  # Remover a coluna total
gender_count_final['total'] = gender_count['total']  # Adicionar a coluna de total de atletas
gender_count_final['percent_female'] = gender_count_percent['Female']
gender_count_final['percent_male'] = gender_count_percent['Male']

# Certificando-se de que ambos os dataframes possuem a coluna 'country' para fazer o join
df_merged = pd.merge(gender_count_final, df_medals_total, on='country',how='left')

# Calculando a razão média de medalhas por atleta
df_merged['avg_medals_per_athlete'] = df_merged['Total'] / df_merged['total']*100
```

# **Narrativa: A Concentração de Medalhas nas Olimpíadas**

As Olimpíadas são o maior evento esportivo do mundo, reunindo atletas de diversas nações para competir pelas medalhas mais cobiçadas. Porém, uma análise detalhada revela uma realidade interessante e que muitas vezes passa despercebida: uma proporção significativa das medalhas conquistadas ao longo da história das Olimpíadas está concentrada em apenas um pequeno grupo de países.

## **Contexto: O Grande Desempenho de Algumas Nações**

Ao observar os resultados de diversas edições dos Jogos Olímpicos, é possível notar que cerca de **50% das medalhas** são conquistadas por **apenas 8 países**. Este dado é impressionante, pois revela como as grandes potências esportivas dominam a maior parte das premiações, enquanto outras nações lutam por uma fração muito menor das medalhas disponíveis.

## **O Impacto da Concentração: O Que Isso Significa?**

Este fenômeno de concentração pode ser explicado por uma série de fatores. Primeiramente, os **investimentos em infraestrutura esportiva**, treinamento e apoio aos atletas são muito mais altos nos países com maior presença nas medalhas. Países como **EUA, China, Rússia, e Japão** possuem programas de treinamento intensivos e recursos abundantes, permitindo que seus atletas se destaquem em uma variedade de modalidades.

Além disso, a **cultura esportiva** e o apoio governamental têm um papel crucial. Em algumas nações, o esporte é visto como uma prioridade nacional, com grande ênfase em competições olímpicas e o desenvolvimento de talentos desde a infância.

## **Visualizando a Concentração: O Papel da Visualização de Dados**

A visualização de dados se torna uma ferramenta essencial para compreender a magnitude dessa concentração. Ao utilizar **gráficos de barras** ou **gráficos de linha**, podemos ilustrar de forma clara como as medalhas estão distribuídas entre os países. Por exemplo, um gráfico de barras pode destacar que mais da metade das medalhas são conquistadas por apenas 8 países, enquanto os outros países, embora participem ativamente, conquistam uma quantidade menor de medalhas.

# Gráfico de Pareto: Análise de Medalhas Olímpicas

O gráfico de Pareto é uma ferramenta poderosa para visualizar a distribuição de dados, permitindo destacar os elementos mais significativos em relação ao total. No contexto de sua análise de medalhas olímpicas, ele mostra como a concentração das medalhas está distribuída entre os países, revelando que uma grande parte das medalhas está concentrada em um número pequeno de países.

## Estrutura do Gráfico de Pareto

O gráfico de Pareto tem duas representações:

- **Barras Verticais (azuis):** Essas barras representam o total de medalhas conquistadas por cada país. As barras são dispostas de maneira decrescente, ou seja, os países com o maior número de medalhas estão à esquerda.

- **Curva Acumulada (vermelha):** A linha vermelha mostra o percentual acumulado das medalhas à medida que você percorre os países. Esse percentual vai de 0% até 100%, e ele revela qual porcentagem do total de medalhas foi conquistada pelos países listados até aquele ponto.

- **Linha Verde (Percentil 50%):** A linha verde horizontal indica o ponto em que 50% do total das medalhas já foram conquistadas. Esse ponto permite entender rapidamente que metade das medalhas está concentrada nos países à esquerda dessa linha.

## Como o Gráfico Ajudou a Chegar à Conclusão

Este gráfico ajudou a perceber que a **metade das medalhas olímpicas** estão concentradas em **apenas 8 países**, como destacado pela linha verde que marca o percentil de 50%. Isso indica que, embora muitos países participem das Olimpíadas, apenas um número muito pequeno de nações domina a competição, com um grande volume de medalhas.

Ao olhar para a **curva acumulada vermelha**, podemos ver que a linha sobe rapidamente nos primeiros países, atingindo 50% do total de medalhas com um pequeno número de nações. Isso significa que há uma alta concentração de medalhas nas potências esportivas, como **EUA**, **China**, **Rússia**, e **Reino Unido**. Depois desse ponto, a curva se estabiliza mais lentamente, mostrando que as medalhas restantes estão mais distribuídas entre os outros países, com uma participação menor.

Essa visualização evidencia a **disparidade no desempenho olímpico** e mostra claramente o domínio de um número restrito de países nas medalhas. Esse é um insight importante que pode ser utilizado para discutir temas como **investimentos em esportes**, **acesso a recursos de treinamento** e **desigualdade entre as nações** nas competições olímpicas.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Supondo que df_medals_total já está carregado
# Ordenar os países pelo número total de medalhas (do maior para o menor)
df_sorted = df_medals_total.sort_values(by='Total', ascending=False)

# Calcular o percentual de medalhas acumuladas
df_sorted['cumulative_percent'] = df_sorted['Total'].cumsum() / df_sorted['Total'].sum() * 100

# Criar o gráfico de Pareto
fig, ax1 = plt.subplots(figsize=(12, 8))

# Gráfico de barras para as medalhas totais
ax1.bar(df_sorted['country'], df_sorted['Total'], color='skyblue', alpha=0.7)
ax1.set_xlabel('País', fontsize=12)
ax1.set_ylabel('Total de Medalhas', fontsize=12)
ax1.tick_params(axis='x', rotation=90)

# Criar o segundo eixo y para a linha cumulativa
ax2 = ax1.twinx()
ax2.plot(df_sorted['country'], df_sorted['cumulative_percent'], color='red', marker='', linestyle='-', linewidth=2)

# Adicionando a linha tracejada no percentil 50
ax2.axhline(y=50, color='green', linestyle='--', linewidth=2, label='Percentil 50%')

# Ajustes adicionais de estilo
ax2.set_ylabel('Percentual Acumulado (%)', fontsize=12)
ax2.set_ylim(0, 100)  # Garantir que o eixo Y do percentil vá de 0 a 100%

# Título e ajustes finais
plt.title('Gráfico de Pareto: Percentual de Medalhas por País', fontsize=14)
fig.tight_layout()

# Adicionando a legenda para a linha tracejada
ax2.legend()

plt.show()
```

# Gráficos Complementares: Distribuição das Medalhas entre Países

Esses gráficos complementam a análise anterior, detalhando a **distribuição das medalhas** entre os **top 8 países** e os **outros países** de uma forma mais visual e intuitiva. Eles foram divididos em duas partes: um gráfico de barras e um gráfico de pizza.

## **1. Gráfico de Barras: Total de Medalhas**

O **gráfico de barras** compara o **total de medalhas** entre o grupo dos **top 8 países** e os **outros países**. Ele apresenta duas barras, uma para cada categoria:

- **Top 8 Países (Azul):** Representa o total de medalhas conquistadas pelos 8 países que mais ganharam medalhas.
- **Outros Países (Laranja):** Representa o total de medalhas conquistadas pelos outros países, aqueles que não estão entre os 8 primeiros.

Neste gráfico, vemos que as barras têm o **mesmo comprimento**, indicando que os **top 8 países** e os **outros países** conquistaram quase **o mesmo número total de medalhas**, refletindo a divisão **50% para cada grupo**.

## **2. Gráfico de Pizza: Distribuição Percentual de Medalhas**

O **gráfico de pizza** apresenta a **distribuição percentual** de medalhas entre os **top 8 países** e os **outros países**. A fatia azul representa **49.7% das medalhas** conquistadas pelos **top 8 países**, enquanto a fatia laranja representa **50.3% das medalhas** conquistadas pelos **outros países**.

Esse gráfico visualiza de maneira clara que a **distribuição de medalhas** está bem equilibrada, com uma **distribuição quase equitativa** entre os dois grupos. Apesar dos **top 8 países** dominarem em termos absolutos de medalhas, a diferença percentual é bem pequena, indicando que os **outros países** também têm uma presença significativa nas medalhas.

## **Conclusão Complementar**

A combinação desses dois gráficos evidencia que, enquanto um pequeno grupo de países domina as medalhas em termos absolutos, a **distribuição percentual** mostra que as outras nações também conquistam uma **quantidade considerável** de medalhas, com uma distribuição **relativamente equilibrada** entre os grupos. Isso complementa a análise do gráfico de Pareto, fornecendo uma **visão clara** da **concentração e distribuição** das medalhas olímpicas.


```python
import pandas as pd
import matplotlib.pyplot as plt

# Ordenar os países pelo número total de medalhas (do maior para o menor)
df_sorted = df_medals_total.sort_values(by='Total', ascending=False)

# Selecionar os Top 8 Países
top_8 = df_sorted.head(8)

# Selecionar os outros países
other_countries = df_sorted.tail(len(df_sorted) - 8)

# Calcular o total de medalhas para Top 8 e Outros Países
total_top_8 = top_8['Total'].sum()
total_other_countries = other_countries['Total'].sum()

# Criar gráfico de barras comparando o total de medalhas entre Top 8 e Outros Países
fig, ax1 = plt.subplots(figsize=(10, 6))

# Plotando as barras com bordas suaves
bars = ax1.bar(['Top 8 Países', 'Outros Países'], [total_top_8, total_other_countries],
               color=['#007bff', '#ff7f0e'], edgecolor='black', linewidth=1.2)

# Adicionando o título e o rótulo do eixo y
ax1.set_title('Total de Medalhas: Top 8 Países vs Outros Países', fontsize=16, fontweight='bold')
ax1.set_ylabel('Total de Medalhas', fontsize=12)
ax1.set_xlabel('Categoria', fontsize=12)

# Ajustando a visualização para um layout mais limpo
ax1.tick_params(axis='x', labelsize=12)
ax1.tick_params(axis='y', labelsize=12)
ax1.set_ylim(0, max(total_top_8, total_other_countries) * 1.1)  # Adicionando um pequeno espaço extra no gráfico

# Exibindo o gráfico de barras
plt.tight_layout()
plt.show()

# Criar gráfico de pizza para mostrar a distribuição percentual
fig2, ax2 = plt.subplots(figsize=(8, 8))

# Adicionando o efeito de "explosão" na fatia do Top 8 Países
ax2.pie([total_top_8, total_other_countries], labels=['Top 8 Países', 'Outros Países'],
        autopct='%1.1f%%', colors=['#007bff', '#ff7f0e'], startangle=90,
        explode=(0.1, 0), wedgeprops={'edgecolor': 'black', 'linewidth': 1.5})

# Título do gráfico de pizza
ax2.set_title('Distribuição Percentual de Medalhas: Top 8 Países vs Outros Países', fontsize=16, fontweight='bold')

# Exibindo o gráfico de pizza
plt.tight_layout()
plt.show()
```
