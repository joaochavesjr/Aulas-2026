```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Gerar dados fictícios
np.random.seed(0)
df = pd.DataFrame({
    'Idade': np.random.randint(20, 60, 100),
    'Renda': np.random.randint(30000, 100000, 100),
    'Educação': np.random.randint(1, 5, 100),
    'Satisfação no Trabalho': np.random.randint(1, 10, 100),
    'Horas de Trabalho Semanais': np.random.randint(20, 60, 100),
    'Índice de Bem-Estar': np.random.randint(1, 10, 100),
    'Nível de Atividade Física': np.random.randint(1, 10, 100),
    'Nível de Estresse': np.random.randint(1, 10, 100)
})

# Calcular a matriz de correlação
correlation_matrix = df.corr()

# Criar o heatmap
plt.figure(figsize=(10, 8))
heatmap = sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5, fmt=".2f")

# Adicionar título ao gráfico
plt.title('Matriz de Correlação - Saúde, Bem-Estar e Trabalho', fontsize=20, pad=20)

# Exibir o gráfico
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Carregar o dataset Iris
iris = sns.load_dataset("iris")

# Separar as features (variáveis independentes) e a variável alvo
X = iris.drop("species", axis=1)
y = iris["species"]

# Padronizar os dados
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Aplicar PCA para reduzir para 2 componentes principais
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Criar um DataFrame com as componentes principais
pca_df = pd.DataFrame(X_pca, columns=["PCA1", "PCA2"])
pca_df["species"] = y

# Plotar o gráfico de dispersão com as componentes principais
plt.figure(figsize=(10, 8))
sns.scatterplot(x="PCA1", y="PCA2", hue="species", data=pca_df, palette="Set2")
plt.title("PCA do Dataset Iris")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Carregar o dataset Iris
iris = sns.load_dataset("iris")

# Separar as features (variáveis independentes)
X = iris.drop("species", axis=1)

# Padronizar os dados
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Aplicar K-Means para encontrar 3 clusters
kmeans = KMeans(n_clusters=3, random_state=0)
clusters = kmeans.fit_predict(X_scaled)

# Adicionar os clusters ao DataFrame
iris['Cluster'] = clusters

# Plotar os clusters com base em duas variáveis principais
plt.figure(figsize=(10, 8))
sns.scatterplot(x="sepal_length", y="sepal_width", hue="Cluster", data=iris, palette="Set1")
plt.title("Clusters no Dataset Iris (K-Means)")
plt.show()
```

```python
# Carregar o dataset Iris
from seaborn import load_dataset
iris = load_dataset('iris')

# Separar as características (features) das classes (target)
X = iris.drop('species', axis=1)  # Características
y = iris['species']  # Classe (target)
```

```python
# Padronizar os dados (isso é importante para o PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Inicializar o PCA para reduzir para 2 componentes principais
pca = PCA(n_components=2)

# Ajustar o PCA aos dados padronizados
X_pca = pca.fit_transform(X_scaled)
```

```python
# Criar um DataFrame para os componentes principais
pca_df = pd.DataFrame(data=X_pca, columns=['PC1', 'PC2'])
pca_df['species'] = y

# Plotar os componentes principais
plt.figure(figsize=(10, 7))
sns.scatterplot(x='PC1', y='PC2', hue='species', data=pca_df, palette='Set2', s=100)
plt.title('PCA do Dataset Iris')
plt.show()
```

### Analise de Cluster com python

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Carregar o dataset Iris
iris = load_iris()
X = iris.data  # Dados das características
y = iris.target  # Rótulos das espécies (não usados para o K-means)
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
kmeans = KMeans(n_clusters=3, n_init=10,random_state=0)

# Ajustar o modelo e prever os clusters
y_kmeans = kmeans.fit_predict(X_scaled)

# Criar um DataFrame para os clusters
cluster_df = pd.DataFrame(X_scaled, columns=iris.feature_names)
cluster_df['Cluster'] = y_kmeans
```

```python
# Plotar os clusters usando as duas primeiras características (Sepal length e Sepal width)
plt.figure(figsize=(10, 7))

sns.scatterplot(x=cluster_df[iris.feature_names[0]],
                y=cluster_df[iris.feature_names[1]],
                hue='Cluster', data=cluster_df,
                palette='Set2',
                s=100)

plt.title('Clusterização com K-means no Dataset Iris')
plt.xlabel('Sepal Length (scaled)')
plt.ylabel('Sepal Width (scaled)')
plt.show()
```

Exercício 1: Matriz de Correlação com Heatmap: Utilize o dataset diamonds para calcular e visualizar a matriz de correlação entre as variáveis numéricas, como carat, depth, table, price, etc. Crie um heatmap dessa matriz de correlação e interprete as principais relações.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Carregar o dataset
diamonds = sns.load_dataset('diamonds')

# Calcular a matriz de correlação
correlation_matrix = diamonds[['carat','depth','table', 'price']].corr()

# Criar o heatmap
plt.figure(figsize=(10,8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Heatmap da Matriz de Correlação das Variáveis Numéricas do Dataset Diamonds')
plt.show()

# Interpretação
# A correlação entre carat e price é geralmente alta, indicando que diamantes maiores
# tendem a ter preços mais elevados. Outras correlações importantes, como depth e price,
# devem ser analisadas para entender sua influência no valor do diamante.
```

Crie um pair plot utilizando o dataset diamonds para visualizar as relações entre as variáveis numéricas. Inclua a variável categórica cut como uma cor diferente nos gráficos para explorar como essa variável influencia as relações entre as variáveis numéricas.

```python
# Criar o pair plot
sns.pairplot(diamonds[['carat','depth','table', 'price','cut']], hue='cut', palette='Set2')
plt.suptitle('Pair Plot das Variáveis Numéricas do Dataset Diamonds com Hue para Cut', y=1.02)
plt.show()

# Interpretação
# O pair plot permite identificar como a variável `cut` influencia a relação entre as outras variáveis,
# como `carat` e `price`. Por exemplo, diamantes de corte "Ideal" podem ter uma distribuição de preço
# diferente em relação aos diamantes de corte "Fair".
```

Exercício 3: Análise de Componentes Principais (PCA)
Reduza a dimensionalidade do dataset diamonds usando PCA para visualizar as principais componentes que explicam a variação nos dados. Concentre-se nas variáveis numéricas e discuta os resultados obtidos.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Selecionar apenas as variáveis numéricas
numeric_data = diamonds.select_dtypes(include=['float64', 'int64'])

# Padronizar os dados
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numeric_data)

# Aplicar PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(scaled_data)

# Variância explicada
explained_variance = pca.explained_variance_ratio_
print("Variância explicada pelos componentes principais:", explained_variance)

# Visualizar os componentes principais
plt.figure(figsize=(8,6))
plt.scatter(pca_result[:, 0], pca_result[:, 1], alpha=0.5)
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.title('PCA aplicado ao Dataset Diamonds')
plt.show()

# Interpretação
# As componentes principais identificadas pelo PCA ajudam a explicar a maior parte da variabilidade dos dados.
# Isso pode ser usado para reduzir a dimensionalidade e facilitar a análise visual e computacional dos dados.
```

Exercício 4: Clusterização com K-means após PCA
Enunciado: Após reduzir a dimensionalidade do dataset diamonds usando PCA, utilize K-means para realizar a clusterização

```python
from sklearn.cluster import KMeans

# Supondo que escolhemos 3 clusters
kmeans = KMeans(n_clusters=3,n_init=11, random_state=42)
clusters = kmeans.fit_predict(pca_result)

# Visualizar os clusters
plt.figure(figsize=(8,6))
plt.scatter(pca_result[:, 0], pca_result[:, 1], c=clusters, cmap='viridis')
plt.xlabel('Componente Principal 1')
plt.ylabel('Componente Principal 2')
plt.title('Clusterização com K-means após PCA no Dataset Diamonds')
plt.show()
```
